```python
workload_file='/home/wuy/DB/pg_mem_data/workloads/tpch_sf1/workload_100k_s1_group_order_by_complex.sql'
with open(workload_file, 'r') as f:
    workload = f.readlines()

```

```python
# find the maximum join query
max_join_query = ''
max_join_count = 0
for query in workload:
    if 'JOIN' in query and 'GROUP' in query and 'ORDER' in query and 'WHERE' in query and 'AND' in query:
        join_count = query.count('JOIN')
        print(f"join_count: {join_count}")
        if join_count > max_join_count:
            max_join_count = join_count
            max_join_query = query

print(max_join_query, 'with max join count', max_join_count)
```
