```python
import torch
import ctypes

# Create a 10GB tensor in RAM
tensor_10GB = torch.randn(int(10 * 1024**3 / 4), dtype=torch.float32, device='cpu')

# Get the pointer to tensor's data and lock it in RAM
addr = tensor_10GB.data_ptr()
size = tensor_10GB.element_size() * tensor_10GB.numel()

# Call mlock to prevent swapping
libc = ctypes.CDLL("libc.so.6")
ret = libc.mlock(addr, size)

if ret == 0:
    print("Successfully locked tensor in RAM, preventing swap usage.")
else:
    print("Failed to lock tensor in RAM. Try running with sudo.")

# Ensure tensor is not released by PyTorch
tensor_10GB.sum().backward() if tensor_10GB.requires_grad else None

print(f"Tensor size: {tensor_10GB.numel() * 4 / 1024**3:.2f} GB")
while 1:pass
```
